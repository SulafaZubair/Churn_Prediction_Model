## DETECTING OUTLIIERS

```python

numeric_cols = data.select_dtypes(include=['int64', 'float64']).columns

for col in numeric_cols:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = max(Q1 - 1.5 * IQR,0) #ignoring negative values
    upper_bound = Q3 + 1.5 * IQR
    
    
    outliers = data[(data[col] < lower_bound) | (data[col] > upper_bound)]
    
   
    print(f"Column: {col}")
    print(f"Number of outliers: {outliers.shape[0]}")
    print(f"Lower bound: {lower_bound}, Upper bound: {upper_bound}")
    print("-"*50)
```

## Behaviour- based Imputation of Rating
The output of .notna() represents a Boolean mask indicating which values are not missing (i.e., which are real/valid values).

True → the value exists, not missing → can use it for calculations

False → missing (NaN) → should be ignored if computing median, mean, etc.




```python
freq_mean = data['order_frequency'].mean()
loyalty_mean = data['loyalty_points'].mean()
```

```python
freq_mean
```

```python
loyalty_mean
```

```python
engaged_median_rating = data[
    (data['order_frequency'] >= freq_mean) &
    (data['loyalty_points'] >= loyalty_mean) &
    (data['rating'].notna())
]['rating'].median()
```

```python
engaged_median_rating
```

High engagement → fill with engaged median

```python
data.loc[
    (data['rating'].isna()) &
    (data['order_frequency'] >= freq_mean) &
    (data['loyalty_points'] >= loyalty_mean),
    'rating'
] = engaged_median_rating
```

```python
data.info()
```

Low engagement → fill with lower quartile rating
Low engagement → fill with low rating

```python
low_rating = data['rating'].quantile(0.25)

data.loc[
    (data['rating'].isna()) &
    (
        (data['order_frequency'] < freq_mean) |
        (data['loyalty_points'] < loyalty_mean)
    ),
    'rating'
] = low_rating
```

Churn is NOT driven by single features

Churn is driven by behavior patterns

Example:

Customer churns when:

low order frequency

AND high delay

AND low loyalty

Each alone may show weak correlation.

But together → strong signal.

This is called:

➡️ Interaction effect

Correlation cannot detect this.

🔹 Real Life Example

Someone doesn’t leave just because:

❌ They are in Karachi
❌ They ordered less once
❌ They used KFC

They leave because of COMBINATIONS like:

✔️ Teenager
✔️ Low loyalty
✔️ Delayed delivery
✔️ Long inactivity

Correlation only checks:

"Does THIS alone affect churn?"

But churn is multi-factor behavior.

🔹 That’s Why Tree Models Work Better

Models like:

🌳 Random Forest
🌳 XGBoost

learn patterns like:

If (low frequency + delay + low loyalty) → churn

Even if individual correlation ≈ 0

🔹 When Correlation is Useful?

Correlation is useful for:

✔️ Finding duplicate features
❌ Not for deciding importance in behavior problems

Churn = behavioral problem

So low correlation is NORMAL ✅

🎯 Final Insight

Low correlation does NOT mean:

❌ Feature is useless

It means:

➡️ Feature alone doesn't explain churn
➡️ But may still help model prediction

## NEW BEHAVIORAL FEATURES

```python
X_train['order_gap'] = X_train['days_since_last_order'] / (X_train['order_frequency'] + 1)
X_test['order_gap'] = X_test['days_since_last_order'] / (X_test['order_frequency'] + 1)

X_train['spend_per_item'] = X_train['price'] / (X_train['quantity'] + 1)
X_test['spend_per_item'] = X_test['price'] / (X_test['quantity'] + 1)

X_train['loyalty_strength'] = X_train['loyalty_points'] / (X_train['active_tenure_days'] + 1)
X_test['loyalty_strength'] = X_test['loyalty_points'] / (X_test['active_tenure_days'] + 1)
```
