```python
import pandas as pd

def add_days_since_last_order(df, reference_date=None):
    """
    Add days_since_last_order feature.

    Parameters:
    df : pandas.DataFrame
        Must contain 'last_order_date' column.
    reference_date : str or pd.Timestamp, optional
        If None, will use the max 'last_order_date' in df.

    Returns:
    df : pandas.DataFrame
        DataFrame with new column 'days_since_last_order'.
    """

    # Ensure datetime
    df['last_order_date'] = pd.to_datetime(df['last_order_date'])

    # Set reference date
    if reference_date is None:
        reference_date = df['last_order_date'].max()
    else:
        reference_date = pd.to_datetime(reference_date)

    # Calculate days since last order
    df['days_since_last_order'] = (reference_date - df['last_order_date']).dt.days
    df['days_since_last_order'] = df['days_since_last_order'].clip(lower=0)

    return df
```
