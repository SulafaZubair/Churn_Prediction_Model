

The `rating` column contains missing values. Simply filling them with mean, median, or mode is not appropriate, because missing ratings themselves carry information about customer engagement and potential churn.


### Rationale

- Missing ratings indicate potential disengagement or dissatisfaction  

## DETECTING OUTLIIERS

```python

numeric_cols = df_data.select_dtypes(include=['int64', 'float64']).columns

# Step 2: Loop through each numeric column and detect outliers
for col in numeric_cols:
    Q1 = df_data[col].quantile(0.25)
    Q3 = df_data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Find outliers
    outliers = df_data[(df_data[col] < lower_bound) | (df_data[col] > upper_bound)]
    
    # Print summary
    print(f"Column: {col}")
    print(f"Number of outliers: {outliers.shape[0]}")
    print(f"Lower bound: {lower_bound}, Upper bound: {upper_bound}")
    print("-"*50)
```
